```python
import papermill as pm
import os   
template_nb = "area2_actpas_auto.ipynb"  
os.makedirs("outputs", exist_ok=True)

monkeys = [
    "Han_20171207",
    "Duncan_20190710", 
]

for monkey in monkeys:
    out_nb = f"results/{monkey}_actpas.ipynb"
    print("Hello from:", monkey, "->", out_nb)

    try:
        pm.execute_notebook(
            template_nb,
            out_nb,
            parameters={"monkey": monkey},
            kernel_name="sdt_env",     
        )
        print("Finished:", out_nb)
    except Exception as e:
        msg = str(e)
        if "Intentional stop (Papermill)" in msg:
            print("Stopped early on purpose for:", monkey)
            continue
        print("FAILED:", monkey,e)
        continue
```
